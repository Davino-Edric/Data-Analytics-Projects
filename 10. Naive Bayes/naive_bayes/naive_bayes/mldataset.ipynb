```python
def hitung_probabilitas(df, kolom_kelas):
    if not isinstance(df, pd.DataFrame):
        raise ValueError("Input harus berupa DataFrame")
    probabilitas = df.drop(columns=[kolom_kelas]).apply(lambda col: df.groupby(kolom_kelas)[col.name].apply(lambda x: (x == 1).mean()))
    return probabilitas
```

```python
hasil_probabilitas = hitung_probabilitas(df, 'label')
print("\nProbabilitas kemunculan tiap nilai dalam setiap kolom berdasarkan kelas:")
hasil_probabilitas
```

```python
def prior(df, kolom_label):
    total_kelas = len(df[kolom_label])
    hasil = {}
    for label, jumlah in df[kolom_label].value_counts().items():
        hasil[label] = jumlah / total_kelas
    return hasil
```

```python
prior_prob = prior(df, 'label')
```

```python
def likelihood(df, kolom_label):
    hasil = {}
    for label in df[kolom_label].unique():
        kelas_df = df[df[kolom_label] == label]
        # Jumlahkan semua kata dalam tiap kelas
        jumlah_kata_dalam_tiap_kelas = kelas_df.drop(kolom_label, axis=1).sum().sum()
        hasil[label] = {}
        for col in df.columns:
            if col != kolom_label:  # Menghindari kolom label
                jumlah_nilai_1 = len(kelas_df[kelas_df[col] == 1])
                hasil[label][col] = jumlah_nilai_1 / jumlah_kata_dalam_tiap_kelas
    return hasil
```

```python
likelihood_prob = likelihood(df, 'label')
likelihood_prob
```

```python
import string
def teks(text):
    lower_text = text.lower()
    # Menghapus semua tanda baca
    no_punctuation_text = lower_text.translate(str.maketrans('', '', string.punctuation))
    # Memisahkan string menjadi daftar kata
    words = no_punctuation_text.split()
    return words
```

```python
def prediksi_kelas(words):
    label_count = {label: 0 for label in df['label'].unique()} #jumlah kemunculan tiap kelas
    for word in words: #iterasi setiap elemen di words
        if word in df.columns: #mengecek apakah word ada sebagai kolom di df
            counts = df[df[word] == 1]['label'].value_counts().to_dict() #{kelas: jumlah kelas yang muncul berdasarkan word}
            for key, value in counts.items(): #loop di setiap key: value di counts
                label_count[key] += value #menambah jumlah kelas jika word ada di kelas tertentu
            print(label_count)
    max_label = max(label_count, key=label_count.get) #menemukan nilai maksimum dalam dictionary label_count
    max_count = label_count[max_label]  #mendapatkan nilai (jumlah kemunculan) dari label yang ditemukan pada langkah sebelumnya (max_label)
    print(f"\nLabel yang paling sering muncul : {max_label} (jumlah kemunculan: {max_count})")
    return max_label
```

```python
import tkinter as tk
from tkinter import messagebox

root = tk.Tk()
root.title("Prediksi Label")
frame = tk.Frame(root)
frame.pack(padx=10, pady=10)
label = tk.Label(frame, text="Input")
label.pack()
entry = tk.Entry(frame, width=60)
entry.pack(pady=(5, 10))

def predict():
    try:
        predicted_label = prediksi_kelas(teks(entry.get())) #sebagai input untuk memprediksi kalimat
        messagebox.showinfo("Prediksi Label", f"Label yang diprediksi: {predicted_label}") #menampilkan prediksi
    except Exception as e:
        messagebox.showerror("Error", f"Terjadi kesalahan: {e}") #menampilkan eror

predict_button = tk.Button(frame, text="Prediksi", command=predict)
predict_button.pack()
root.mainloop()
```
