```python
import random

def add_noise_and_save(df, nan_fraction=0.1, dup_fraction=0.1, output_file='coffee_shop_revenue_noisy.csv'):
    """
    Randomly add NaN values and duplicate rows to the DataFrame, then save to CSV.
    
    Parameters:
    - df: pandas DataFrame
    - nan_fraction: fraction of cells to set to NaN (default 0.1)
    - dup_fraction: fraction of rows to duplicate (default 0.1)
    - output_file: output CSV file path (default 'coffee_shop_revenue_noisy.csv')
    """
    # Make a copy to avoid modifying original
    noisy_df = df.copy()
    
    # Add NaN values randomly
    total_cells = noisy_df.size
    nan_count = int(total_cells * nan_fraction)
    for _ in range(nan_count):
        row_idx = random.randint(0, len(noisy_df) - 1)
        col_idx = random.randint(0, len(noisy_df.columns) - 1)
        noisy_df.iloc[row_idx, col_idx] = np.nan
    
    # Duplicate rows randomly
    dup_count = int(len(noisy_df) * dup_fraction)
    dup_indices = random.sample(range(len(noisy_df)), dup_count)
    duplicated_rows = noisy_df.iloc[dup_indices]
    noisy_df = pd.concat([noisy_df, duplicated_rows], ignore_index=True)
    
    # Save to CSV
    noisy_df.to_csv(output_file, index=False)
    print(f"Noisy data saved to {output_file}")
    return noisy_df

# Apply the function to df
noisy_df = add_noise_and_save(df)
```
