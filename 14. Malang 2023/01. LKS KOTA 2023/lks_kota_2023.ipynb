```python
np.random.seed(0)
train_size = int(len(x) * 0.8)
idx = np.random.permutation(len(x))

x_train, x_test = x.iloc[idx[:train_size]], x.iloc[idx[train_size:]]
y_train, y_test = y.iloc[idx[:train_size]], y.iloc[idx[train_size:]]

print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
def normalisasi(train_fitur):
    min_value = train_fitur.min(axis=0)
    max_value = train_fitur.max(axis=0)
    norm_value = (train_fitur - min_value) / (max_value - min_value)
    return norm_value, min_value, max_value
```

```python
x_train, min_value, max_value = normalisasi(x_train)
```

```python
x_test = (x_test - min_value) / (max_value - min_value)
```

```python
class SimpleLinearRegression:
    def __init__(self):
        self.slope = None
        self.intercept = None

    def fit(self, x, y):
        n = len(x)
        numerator = n * np.sum(x * y) - np.sum(x) * np.sum(y)
        denominator = n * np.sum(x**2) - np.sum(x)**2
        self.slope = numerator / denominator
        self.intercept = (np.sum(y) - self.slope * np.sum(x)) / n

    def predict(self, x):
        y_pred = self.intercept + self.slope * x
        return y_pred
```

```python
model = SimpleLinearRegression()
```

```python
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
def evaluate(y_true, y_pred):
    r2 = 1 - (np.sum((y_true - y_pred)**2) / np.sum((y_true - y_true.mean())**2))
    mse = np.mean((y_true - y_pred)**2)
    mae = np.mean(np.abs(y_true - y_pred))
    return {'R-Squred': r2, 'Mean Squared Error': mse, 'Mean Absolute Error': mae}
```

```python
for key, value in evaluate(y_test, y_pred).items():
    print(f'{key}: {value}')
```

```python
def normalisasi_input(input_data, min_value, max_value):
    return (input_data - min_value) / (max_value - min_value)
```
